Purpose: To perform checks on the data for a better chance at creating a robust algorithm

```python
import pandas as pd
import json
```

```python
valid_records = []
with open("payments.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:  # Skip empty lines
            continue
        try:
            valid_records.append(json.loads(line))
        except json.JSONDecodeError:
            continue  # Skips the malformed rows
```

```python
df = pd.DataFrame(valid_records)
df.head()
```

```python
df_with_same_id = df[df.duplicated(subset=['txn_id'], keep="first")]
df[df["txn_id"] == "txn_007794_1059"]
```
